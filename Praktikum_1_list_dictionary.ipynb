## 1. Operasi fundamental List
`list` bersifat terurut, dapat diubah, dan bisa berisi banyak jenis data. Contoh berikut membahas indexing, append, insert, extend, delete, iteration, sorting, copying, aggregation, joining, dan type checking.

```python
angka = [10, 20, 30, 40, 50]
print("Elemen pertama:", angka[0])
print("Elemen terakhir:", angka[-1])
print("Irisan 1:4:", angka[1:4])
print("Lompat dua:", angka[::2])
```

```python
angka.append(60)
angka.insert(2, 25)
angka.extend([70, 80, 90])
print("Setelah tambah:", angka)
terhapus = angka.pop(3)
angka.remove(25)
del angka[0:2]
print("Elemen yang di-pop:", terhapus)
print("Setelah hapus:", angka)
```

```python
for indeks, nilai in enumerate(angka):
    print(f"Indeks {indeks}: {nilai}")
kuadrat = [nilai**2 for nilai in angka if nilai > 30]
print("Kuadrat nilai > 30:", kuadrat)
print("Urutan naik:", sorted(angka))
print("Urutan turun:", sorted(angka, reverse=True))
```

```python
asli = [1, 2, 3]
alias = asli
salinan = asli.copy()
asli.append(4)
print("Alias berubah:", alias)
print("Salinan tetap:", salinan)

nilai = [10, 20, 30, 40, 50]
print("Min, max, jumlah, rata-rata:", min(nilai), max(nilai), sum(nilai), sum(nilai)/len(nilai))
print("Gabungan:", [1, 2, 3] + [4, 5])
print("Tipe campuran:", [type(x).__name__ for x in [1, 'halo', 3.14, True]])
```

## 2. Operasi fundamental Dictionary
`dict` menyimpan informasi sebagai pasangan `key: value` sehingga cocok untuk data mahasiswa atau dataset berlabel.

```python
mahasiswa = {
    'nama': 'Alya', 'umur': 20,
    'nilai': [85, 90, 78], 'mata_kuliah': {'Matematika': 'A'}
}
print('Nama:', mahasiswa['nama'])
print('Alamat (default):', mahasiswa.get('alamat', 'Tidak ada'))
mahasiswa['jurusan'] = 'Sistem Informasi'
mahasiswa.update({'umur': 21, 'semester': 4})
print('Setelah pembaruan:', mahasiswa)
```

```python
nilai_dihapus = mahasiswa.pop('nilai')
del mahasiswa['mata_kuliah']
print('Data nilai yang dihapus:', nilai_dihapus)
for k, v in mahasiswa.items():
    print(f'{k}: {v}')
salinan_mahasiswa = mahasiswa.copy()
kuadrat_dict = {n: n*n for n in range(1, 6)}
print('Dictionary comprehension:', kuadrat_dict)
print('Merge:', {**{'a': 1, 'b': 2}, **{'b': 3, 'c': 4}})
```

```python
nilai_mapel = {'Matematika': 85, 'Fisika': 92, 'Kimia': 78}
print('Rata-rata nilai:', sum(nilai_mapel.values())/len(nilai_mapel))
sekolah = {'siswa': {'Alya': {'umur': 20, 'nilai': 'A'}, 'Beni': {'umur': 21, 'nilai': 'B'}}}
sekolah['siswa']['Alya']['nilai'] = 'A+'
print('Nested dictionary:', sekolah['siswa']['Alya'])
```

## 3. Latihan 1
**TODO 1:** filter suhu pada rentang 22–25 °C. **TODO 2:** normalisasi min–max 0–1. **TODO 3:** moving average *window* 3. **TODO 4–5:** pengelompokan label dan rata-rata tiap fitur per kelas.

```python
# Data mentah dari latihan dosen
raw_data = [23.5, 24.1, 22.8, 25.3, 21.9, 26.7, 20.5, 24.8, 23.2, 25.9]

# TODO 1 — seleksi rentang inklusif
filtered_data = [suhu for suhu in raw_data if 22 <= suhu <= 25]

# TODO 2 — min-max scaling
minimum, maksimum = min(raw_data), max(raw_data)
selisih = maksimum - minimum
normalized_data = [(suhu - minimum) / selisih if selisih else 0.0 for suhu in raw_data]

# TODO 3 — moving average sederhana
def moving_average(data, window_size):
    if not isinstance(window_size, int) or not (1 <= window_size <= len(data)):
        raise ValueError('window_size harus bilangan bulat antara 1 dan panjang data')
    return [sum(data[i:i + window_size]) / window_size
            for i in range(len(data) - window_size + 1)]

ma_result = moving_average(raw_data, 3)
print('Data terfilter:', filtered_data)
print('Data normalisasi:', [round(x, 3) for x in normalized_data])
print('Moving average:', [round(x, 3) for x in ma_result])
```

```python
dataset = [
    {'features': [1.2, 3.4, 2.1], 'label': 'class_A'},
    {'features': [2.3, 1.5, 4.2], 'label': 'class_B'},
    {'features': [3.1, 2.8, 1.9], 'label': 'class_A'},
    {'features': [4.2, 3.9, 2.5], 'label': 'class_B'}
]

# TODO 4 — kelompokkan fitur menurut label
def group_by_label(data):
    kelompok = {}
    for baris in data:
        kelompok.setdefault(baris['label'], []).append(baris['features'])
    return kelompok

# TODO 5 — rata-rata setiap dimensi fitur
def average_features_per_class(grouped_data):
    hasil = {}
    for label, daftar_fitur in grouped_data.items():
        if not daftar_fitur:
            hasil[label] = []
            continue
        hasil[label] = [sum(kolom) / len(kolom) for kolom in zip(*daftar_fitur)]
    return hasil

grouped_data = group_by_label(dataset)
avg_features = average_features_per_class(grouped_data)
print('Kelompok data:', grouped_data)
print('Rata-rata fitur per kelas:', avg_features)
```

```python
# Pengujian untuk memastikan solusi memenuhi tujuan latihan
assert filtered_data == [23.5, 24.1, 22.8, 24.8, 23.2]
assert len(normalized_data) == len(raw_data)
assert all(0 <= x <= 1 for x in normalized_data)
assert len(ma_result) == len(raw_data)-2
assert set(grouped_data) == {'class_A', 'class_B'}
assert len(avg_features['class_A']) == 3
assert abs(avg_features['class_A'][0] - 2.15) < 1e-10
print('✅ Praktikum 1 dan semua latihan berhasil.')
```
