## 3. Latihan 2 
**TODO 1–4:** Z-score, penanganan outlier, *one-hot encoding*, dan *train-test split* manual tanpa scikit-learn.

```python
# Dataset latihan dosen: 100 sampel x 5 fitur
np.random.seed(42)
X = np.random.randn(100, 5) * 10 + 5

# TODO 1 — Z-score tiap fitur
def z_score_normalization(data):
    array = np.asarray(data, dtype=float)
    rata = np.mean(array, axis=0)
    simpangan = np.std(array, axis=0)
    simpangan_aman = np.where(simpangan == 0, 1.0, simpangan)
    return (array - rata) / simpangan_aman

X_normalized = z_score_normalization(X)
print('Rata-rata per kolom:', np.round(X_normalized.mean(axis=0), 8))
print('Std per kolom:', np.round(X_normalized.std(axis=0), 8))
```

```python
# TODO 2 — batas outlier = mean +/- threshold x std (per fitur)
def handle_outliers(data, std_threshold=3):
    array = np.asarray(data, dtype=float)
    rata = np.mean(array, axis=0)
    std = np.std(array, axis=0)
    batas_bawah = rata - std_threshold * std
    batas_atas = rata + std_threshold * std
    return np.clip(array, batas_bawah, batas_atas)

X_cleaned = handle_outliers(X_normalized)
print('Bentuk data setelah clipping:', X_cleaned.shape)
```

```python
# TODO 3 — one-hot encoding label kategorikal, termasuk label non-berurutan
def one_hot_encoding(labels):
    labels = np.asarray(labels)
    kelas, indeks = np.unique(labels, return_inverse=True)
    return np.eye(len(kelas), dtype=int)[indeks]

labels = np.array([0, 1, 2, 0, 1, 2, 0])
one_hot_labels = one_hot_encoding(labels)
print('One-hot encoding:\n', one_hot_labels)
```

```python
# TODO 4 — pemisahan data tanpa sklearn
def train_test_split_numpy(X, y, test_size=0.2):
    X, y = np.asarray(X), np.asarray(y)
    if len(X) != len(y):
        raise ValueError('Jumlah fitur dan label harus sama')
    if not 0 < test_size < 1:
        raise ValueError('test_size harus di antara 0 dan 1')
    indeks = np.random.default_rng(42).permutation(len(X))
    n_test = max(1, int(np.ceil(len(X) * test_size)))
    indeks_test, indeks_train = indeks[:n_test], indeks[n_test:]
    return X[indeks_train], X[indeks_test], y[indeks_train], y[indeks_test]

y = np.random.randint(0, 3, 100)
X_train, X_test, y_train, y_test = train_test_split_numpy(X, y)
print('Train:', X_train.shape, y_train.shape)
print('Test:', X_test.shape, y_test.shape)
```
